```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Original XOR dataset in 2D
X = np.array([[1,1],[-1,-1],[1,-1],[-1,1]])
y = np.array([1,1,-1,-1])


# Plotting the 2D points
plt.figure(figsize=(8,4))
for i,label in enumerate(y):
    plt.plot(X[i,0],X[i,1],'ro' if label == 1 else 'bo')
plt.xlabel('x1')
plt.ylabel('y1')
plt.title('Original XOR points in 2D')
plt.axhline(0,color='black',linewidth=0.5)
plt.axhline(0,color='black',linewidth=0.5)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

X = np.array([[1,1],[-1,-1],[1,-1],[-1,1]])
y = np.array([1,1,-1,-1])
X_transformed = np.hstack((X, (X[:,0] * X[:,1]).reshape(-1,1)))

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

colors = ['red' if label == 1 else 'blue' for label in y]
ax.scatter(X_transformed[:,0], X_transformed[:,1], X_transformed[:,2], 
           c=colors, s=100, depthshade=True)

xx, yy = np.meshgrid(np.linspace(-2,2,50), np.linspace(-2,2,50))
zz = np.zeros_like(xx)
ax.plot_surface(xx, yy, zz, alpha=0.5, color='green')

ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.set_zlabel('x1*x2')
ax.set_title('XOR Problem in 3D Space')
plt.tight_layout()
plt.show()
```
